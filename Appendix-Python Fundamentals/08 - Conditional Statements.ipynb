<h1 style="text-align: justify;">Conditional Statements</h1>
<p style="text-align: justify;">


Conditional statements are control structures that allow blocks of code to be executed based on logical conditions, using if, else, and elif in Python.
    
</p>
</div>

```python
# Example 1: Print a message based on the time
current_hour = 15

if 5 <= current_hour < 12:
    message = "Good morning"
elif 12 <= current_hour < 18:
    message = "Good afternoon"
else:
    message = "Good evening"
    
print(message)
```

```python
# Example 2: Simulate the temperature state in degrees Celsius
temperature = 28

if temperature > 30:
    state = "hot"
elif 20 <= temperature <= 30:
    state = "pleasant"
else:
    state = "inconsistent"
    
print(state)
```

```python
# Example 3: Simulate weather conditions and time of day
is_day = False
is_rainy = True

# Evaluate the weather conditions
if is_day:
    if is_rainy:
        print("Do not leave home, or remember to take your umbrella")
    else:
        print("It is a wonderful day")
else:
    print("You can watch the stars")
```
