In Python, a class is a blueprint for creating objects that group related data and functions. It defines the structure and behavior of objects. Classes enable code reuse and the implementation of object-oriented programming.

</p>
</div>

```python
class Car:
    
    """
    Class that represents an abstraction of a real-world car.
    """
    
    def __init__(self, brand, model, color):
        
        # Define attributes
        self.brand = brand
        self.model = model
        self.color = color
        self.is_on = False
        self.speed = 0
        
    def start(self):
        if not self.is_on:
            print(f"The {self.brand} {self.model} {self.color} has been started")
            self.is_on = True
        else:
            print("The car is already running")
            
    def turn_off(self):
        if self.is_on:
            print(f"The {self.brand} {self.model} {self.color} has been turned off")
            self.is_on = False
            self.speed = 0
        else:
            print("The car is already off")
    
    def accelerate(self, speed):
        if self.is_on:
            self.speed += speed
            print(f"The {self.brand} {self.model} {self.color} accelerates to {self.speed} km/h")
        else:
            print("Start the car first!")
```

```python
# Create an instance of the Car class
car1 = Car(brand="Toyota", model="Corolla", color="White")

# Interact with our car
car1.start()
car1.accelerate(50)
car1.turn_off()
```

```python
# Create an instance of the Car class
car2 = Car(brand="Ford", model="Fiesta", color="Black")

# Interact with our car
car2.start()
car2.accelerate(50)
car2.turn_off()
```
