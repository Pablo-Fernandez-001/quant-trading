<h1 style="text-align: justify;">Dictionaries</h1>
<p style="text-align: justify;">

Dictionaries in Python are collections of elements stored as key-value pairs. They allow data to be stored and accessed through unique keys, making information management efficient. Dictionaries are mutable, dynamic, and can contain values of different data types.
    
</p>
</div>

```python
# Create a dictionary
person_information = {
    "name": "John",
    "age": 30,
    "country": "Costa Rica"
}

# Access values using keys
person_name = person_information["name"]
person_age = person_information["age"]

# Print the values
print("Name:", person_name)
print("Age:", person_age)
```

### Clear

```python
# Create a dictionary with data about books and their genres.
books = {"Harry Potter": "Fantasy", "1984": "Science Fiction", "One Hundred Years of Solitude": "Magical Realism"}

# Clear the dictionary to remove the genre information.
books.clear()

# Print the empty dictionary after removing the genre information.
print(books)
```

### Copy


```python
# Create a dictionary with data about fruits and their colors.
fruit_colors = {"Apple": "Red", "Banana": "Yellow", "Grape": "Purple"}

# Create a copy of the fruit and color dictionary.
fruit_colors_copy = fruit_colors.copy()

# Print the copy of the fruit and color dictionary.
print(fruit_colors_copy)
```

### Fromkeys

```python
# Create a list of attributes for a vehicle.
vehicle_attributes = ["brand", "model", "color"]
default_value = "Not specified"

# Create a vehicle attribute dictionary with default values.
vehicle_dict = dict.fromkeys(vehicle_attributes, default_value)

# Print the vehicle attribute dictionary with default values.
print(vehicle_dict)
```

### Get

```python
# Create a dictionary with information about countries and their capitals.
country_capitals = {"United States": "Washington D.C.", "France": "Paris", "Japan": "Tokyo"}

# Get the capital of France using the get method.
france_capital = country_capitals.get("France", "Not specified")

# Print the capital of France.
print(france_capital)
```

### Keys

```python
# Create a dictionary with information about subjects and their codes.
subject_codes = {"Mathematics": "MAT101", "History": "HIS201", "English": "ENG301"}

# Get a view of the dictionary keys (subjects).
subject_keys = subject_codes.keys()

# Print the key view (subjects).
print(subject_keys)
```

### Pop

```python
# Create a dictionary with information about foods and their calories.
food_calories = {"Apple": 95, "Banana": 105, "Orange": 62}

# Remove the Banana calorie information using the pop method.
banana_calories = food_calories.pop("Banana")
print(banana_calories) # Returns only the value

# Print the dictionary after removing the Banana calorie information.
print(food_calories)
```

### Popitem

```python
# Create a dictionary with information about planets and their distances from the Sun.
planet_distances = {'Mercury': 57.9, 'Venus': 108.2, 'Mars': 227.9}

# Remove and return a key-value pair using the popitem method.
removed_planet = planet_distances.popitem() # Removes the last value

# Print the key-value pair removed from the dictionary.
print(removed_planet)
print(planet_distances)
```

### Setdefault

```python
# Create a dictionary with information about colors and their hexadecimal codes.
color_codes = {'Red': '#FF0000', 'Green': '#00FF00'}

# Insert the hexadecimal code for Blue using the setdefault method.
blue_code = color_codes.setdefault('Blue', '#0000FF')

# Print the dictionary after adding Blue with its hexadecimal code.
print(color_codes)
```

### Update

```python
# Create a dictionary with information about tasks and their statuses.
task_statuses = {'Task1': 'Pending', 'Task2': 'In progress'}

# Update the task statuses using the update method.
new_statuses = {'Task1': 'Completed', 'Task3': 'Pending'}
task_statuses.update(new_statuses)

# Print the dictionary after updating the statuses.
print(task_statuses)
```

### Values

```python
# Create a dictionary with information about products and their prices.
product_prices = {'Laptop': 1200, 'Phone': 600, 'Tablet': 300}

# Get a view of the product prices using the values method.
product_prices_view = product_prices.values()

# Print the product price view.
print(product_prices_view)
```

### Items

```python
# Create a dictionary with information about movies and their release years.
movie_years = {"Titanic": 1997, "Matrix": 1999, "The Godfather": 1972}

# Get a view of tuples representing movie-year pairs.
movie_info = movie_years.items()

# Print the movie-year pair view.
print(movie_info)
```
