### Integer

```python
# Positive Number
positive_number = 100
positive_number
```

```python
# Large Positive Number
large_number = 1_000_000
large_number
```

```python
# Negative Number
negative_number = -50
negative_number
```

```python
# Operations
addition = positive_number + negative_number
subtraction = positive_number - negative_number
division = positive_number / negative_number
multiplication = positive_number * negative_number

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Division:", division)
print("Multiplication:", multiplication)
```

```python
# Check the data type
reference = 10
type(reference)
```

```python
# Integer methods
dir(int)
```

```python
# Apply some methods
integer_example = 5
integer_example.to_bytes(), integer_example.denominator, integer_example.numerator
```

### Float

```python
# Floating-point number
pi = 3.141592
pi
```

```python
# Negative floating-point number
negative_float_number = -106.25
negative_float_number
```

```python
# Operations
addition = pi + negative_float_number
subtraction = pi - negative_float_number
division = pi / negative_float_number
multiplication = pi * negative_float_number

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Division:", division)
print("Multiplication:", multiplication)
```

```python
# Operations between integers and floats
integer = 10
floating = 50.345
result = integer + floating
result
```

```python
# Convert one data type to another
integer_result = int(result)
integer_result, type(integer_result)
```

```python
# Float
float_result = float(integer_result)
float_result, type(float_result)
```

```python
# Available methods
dir(float)
```

```python
# Apply methods
float_example = 15.24
float_example.is_integer(), float_example.hex()
```

### Boolean

```python
# Data types
true_value = True
false_value = False
true_value, false_value
```

```python
result = 5 > 3
result
```

```python
# Available methods
dir(bool)
```

```python
# Apply methods
true_value.to_bytes(), true_value.conjugate()
```

### Strings

```python
# Examples
simple_string = 'Hello, Python'
double_quoted_string = "Hello, Python"
triple_quoted_string = """
This is an example of triple quotes with 
multiple lines.
"""
```

```python
simple_string, double_quoted_string, triple_quoted_string
```

```python
# Concatenate
name = "John"
last_name = "Perez"

full_name = name + " " + last_name
full_name
```

```python
# Available methods
dir(str)
```

```python
text = "Hello, how are you?"
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Replace:", text.replace("Hello", "Helloooo"))
```
