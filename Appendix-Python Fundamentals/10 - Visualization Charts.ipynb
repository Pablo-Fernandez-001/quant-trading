<h1 style="text-align: justify;">Visualization Charts</h1>
<p style="text-align: justify;">

Visualization charts are visual representations of data that help identify patterns, trends, and relationships. In Python, they can be created using libraries such as Matplotlib, Seaborn, Plotly, and others, which provide a wide range of options for creating interactive and static charts.

### Matplotlib

```python
import matplotlib.pyplot as plt

# Line chart
x = [1, 2, 3, 4, 5]
y = [1, 2, 3, 4, 5]
plt.plot(x, y)
plt.grid()
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Line Chart")
plt.show()
```

```python
import numpy as np

x = np.random.random(size=(100,))
y = np.random.random(size=(100,))
plt.scatter(x, y)
plt.grid()
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Scatter Plot")
plt.show()
```

### Seaborn

```python
import seaborn as sns
import pandas as pd
from sklearn.datasets import load_iris

# Load the data
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df["species"] = pd.Categorical.from_codes(iris.target, iris.target_names)
iris_df
```

```python
# Create chart
sns.scatterplot(x="sepal length (cm)", y="sepal width (cm)", hue="species", data=iris_df)
plt.title("Scatter Plot - Iris Dataset")
plt.show()
```

### Pandas

```python
data = np.random.normal(size=(100_000,))
df = pd.DataFrame(data)
df.plot(kind="hist", bins=100)
plt.grid()
plt.title("Histogram with Pandas")
plt.show()
```

### Alternatives:

There are many libraries available for creating charts. The most popular are Matplotlib and Seaborn.

Optional libraries:


- Plotly: Allows the creation of interactive charts and dashboards. It is especially useful for web visualizations and supports a wide variety of chart types.

- Bokeh: Provides an elegant way to build interactive charts and visual web applications. It is especially useful when interaction with charts in a web browser is required.

- Altair: A declarative library that allows interactive charts to be created using a simple and concise syntax. It integrates with Pandas and is easy to learn.

- Dash (by Plotly): A library for creating interactive Python-based web applications. It allows interactive dashboards to be built.

- ggplot (by yhat): Based on R's popular ggplot2 graphics system, it provides a similar interface for creating grammar-of-graphics visualizations in Python.

- NetworkX: Specialized in the visualization and analysis of complex graphs and networks. It is useful for representing relationships and network structures.

- Holoviews: A library that simplifies data exploration and visualization by providing a simple abstraction for complex charts.
