<h1 style="text-align: justify;">Exception Handling</h1>
<p style="text-align: justify;">

In Python, an exception is an event that occurs during program execution and interrupts the normal flow of execution. Exceptions can be caused by errors in the code or by exceptional conditions that arise while the program is running.

Exception handling in Python allows these errors to be managed using try-except blocks. The try block is used to execute code that may raise an exception, while the except block is used to handle the exception if it occurs.
    
</p>
</div>

### TypeError

This error occurs when an operation or function is applied to data of an inappropriate type.

```python
try:
    addition = '10' + 5
except TypeError as e: #- At this point we are assigning the exception object to a variable named e
    print(f"Incompatible data type error! Error -> {e}")
```

### ZeroDivisionError

This error occurs when attempting to divide by zero, which is mathematically undefined.

```python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"Division by zero error! Error -> {e}")
```

### OverflowError

This error occurs when a calculation exceeds the limit supported by a numeric data type. It is not very common, but we can create an example.

```python
try:
    calculation = 5.25 ** 25665
except OverflowError as e:
    print(f"Numeric overflow error! Error -> {e}")
```

### ValueError

This error occurs when an operation or function receives an argument with an inappropriate value.

```python
try:
    number_value = int('abc')  #- Raises a ValueError when attempting to convert a non-numeric string to an integer.
except ValueError as e:
    print(f"Invalid value error! Error -> {e}")
```

### IndexError

This error occurs when the index of a sequence, whether a tuple, list, or string, is out of range. For example, trying to access index 60 in a list containing only 5 elements.

```python
try:
    example_list = [1, 2, 3]
    element = example_list[10]
except IndexError as e:
    print(f"Index out of range error! Error ->  {e}")
```

### NameError

This error occurs when we call a function or method, or try to access a variable that does not exist.

```python
try:
    print(undefined_variable)  #- Raises a NameError when attempting to access an undefined variable.
except NameError as e:
    print(f"Undefined name error! Error -> {e}")
```

### ModuleNotFoundError

This error occurs when an import statement attempts to load a module that does not exist or cannot be found.

```python
try:
    import nonexistent_module
except ImportError as e:
    print(f"Module import error! Error -> {e}")
```

### KeyError

This error occurs when a key cannot be found in a dictionary.

```python
try:
    dictionary = {"a": 1, "b": 2}
    value = dictionary['c']
except KeyError as e:
    print(f"Missing dictionary key error! Error ->  {e}") #- Only prints
```

### FileNotFoundError

This error occurs when we try to open a file that does not exist.

```python
try:
    with open("file_that_does_not_exist.txt", "r") as file:
        content = file.read()
except FileNotFoundError as e:
    print(f"File not found error! Error ->  {e}")
```

### Exception

This catches errors regardless of their specific exception type.

```python
try:
    "ABC" + 10
except Exception as e:
    print(f"Execution error: {e}")
```

### Multiple Errors in a Single Block

```python
try: 
    "ABC" + 15
except ZeroDivisionError as e:
    print(f"Error ZeroDivisionError: {e}")
except (OverflowError, IndexError, FileNotFoundError):
    print("Error of any type (TypeError, IndexError, FileNotFoundError): {e}")
except Exception as e:
    print(f"Execution error: {e}")
```
