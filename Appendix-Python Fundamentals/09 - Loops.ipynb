<h1 style="text-align: justify;">Loops</h1>
<p style="text-align: justify;">

Loops are control structures that allow a block of code to be repeated multiple times. In Python, the most common loops are the for loop and the while loop. They allow a set of instructions to be executed repeatedly while a condition remains true or until a collection of elements has been exhausted.

    
</p>
</div>

### For Loops

```python
# Example 1: Iterate over a list
fruits = ["apple", "banana", "cherry"]
for i in fruits:
    print(i)
```

```python
# Example 2: Print numbers within a range
for number in range(0, 5 + 1, 1):
    print(number)
```

```python
# Example 3: Iterate over a word
message = "Test Word"
for letter in message:
    print(letter)
```

```python
# Example 4: Calculate the sum of the values in a range
addition = 0
for number in range(1, 1_000):
    addition += number
    
print("The sum is:", addition)
```

```python
# Example 5: Iterate over a dictionary
students = {"John": 25, "Ana": 22, "Pedro": 28}
for name, age in students.items():
    print(f"{name} is {age} years old")
```

### While Loops

```python
# Example 1: Decreasing counter
counter = 5
while counter > 0:
    print(counter)
    counter -= 1
```

```python
# Example 2: Add numbers up to a limit
limit = 10
addition = 0
while addition <= limit:
    print(addition)
    addition += 2
```

```python
# Example 3: Search for an element
target_element = 8
example_list = [1, 3, 5, 7, 9]
index = 0
while (index < len(example_list)) and (example_list[index] != target_element):
    index += 1
    
if index < len(example_list):
    print(f"The element was found at index {index}")
else:
    print(f"{target_element} is not in the list")
```
