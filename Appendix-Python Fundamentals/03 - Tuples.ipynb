<h1 style="text-align: justify;">Tuples</h1>
<p style="text-align: justify;">

A tuple in Python is similar to a list, but it is immutable, which means its elements cannot be modified after creation. It is defined using parentheses (1, 2, 3), and each element must be separated by commas.
    
</p>
</div>

```python
# Create a tuple
my_tuple = (1, 2, 3, "four", 5.9)
my_tuple
```

```python
# Data type
type(my_tuple)
```

```python
# Access tuple elements
my_tuple[0], my_tuple[-2]
```

```python
# Unpack
my_tuple = (1, "two", 3.0, (4, 5))

a, b, c, d = my_tuple
a, b, c, d
```

```python
d
```

```python
# Add tuples: This also applies to lists
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
new_tuple = tuple1 + tuple2
new_tuple
```

```python
# Available methods
dir(tuple)
```

### Count

```python
# Occurrences
example_tuple = (1, 2, 3, 1, 1, 1)
occurrences = example_tuple.count(1)
occurrences
```

### Index

```python
# Find the first occurrence of an element
example_tuple = (1, 2, 3, 1, 1, 1)
first_occurrence = example_tuple.index(1)
first_occurrence
```
