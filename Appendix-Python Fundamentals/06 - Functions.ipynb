<h1 style="text-align: justify;">Functions</h1>
<p style="text-align: justify;">

Functions in Python are reusable blocks of code that perform a specific task. They help modularize code and improve readability and maintainability. Functions are defined with the "def" keyword, can receive parameters, and can return values.
    
</p>
</div>

```python
# Define example
def greet(name: str) -> None:
    
    """
    Print a greeting to the console
    """
    
    print(f"Hello, {name}!")
    
# Call the function
greet("Andrea")
```

```python
def greet_country(name: str, country: str = "France") -> None:
    
    """
    Print a greeting to the console
    """
    
    print(f"Hello, {name}! Greetings from {country}")
    
# Call the function
greet_country("Roberto")
```

```python
def square_root(x: float) -> float:
    
    """
    Calculates the square root of a value
    """
    
    return x**(1/2)

# Call the function
square_root(9)
```

### Lambda Functions

These are anonymous functions, defined in a single line, for simple tasks.

```python
concatenate_sentences = lambda x, y: x + " " + y
concatenate_sentences("Luis", "Rodriguez")
```
