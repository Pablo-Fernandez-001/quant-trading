<h1 style="text-align: justify;">Lists</h1>
<p style="text-align: justify;">

A list in Python is a data structure that can contain elements of any type, such as numbers, strings, booleans, and more. It is defined using square brackets ["A", 2, [1, 2, 3], 4.0] with elements separated by commas. Lists are mutable, which means they can be modified by adding, removing, or changing elements.
    
</p>
</div>

```python
# Create a list
my_list = [1, 2, 3, "four", 5.0]
my_list
```

```python
# Data type
type(my_list)
```

```python
# Access list elements
first_element = my_list[0]
last_element = my_list[-1]
first_element, last_element
```

```python
# Modify list elements
my_list[1] = "two"
my_list
```

```python
# Available methods
dir(list)
```

### Append

```python
# Example
example_list = [1, 2, 3, 4, 5]
example_list
```

```python
# Add a value
example_list.append(6)
example_list
```

```python
# Add another list
example_list.append([7, 8, 9])
example_list
```

### Clear


```python
print(example_list)
example_list.clear()
print(example_list)
```

### Copy

```python
# Copy a list
my_list = [1, 2, 3]
copied_list = my_list.copy()
print(copied_list)
```

```python
# Disadvantages of not copying a list
original_list = [1, 2, 3, 4]
list_copy = original_list
original_list[1] = "Hello"
print(original_list)
print(list_copy)
```

```python
# Advantages of using the copy method
original_list = [1, 2, 3, 4]
list_copy = original_list.copy()
original_list[1] = "Hello"
print(original_list)
print(list_copy)
```

### Count

```python
# Count
my_list = [1, 2, 2, 3, 4, 5, 2, 6, 7, 8, 9]
count_2 = my_list.count(2)
print(count_2)
```

### Extend

```python
# Example
example_list = [1, 2, 3, 4, 5]
example_list
```

```python
# Extend
example_list.extend([7, 8, 9])
example_list
```

### Index

```python
# Get the first occurrence
my_list = [1, 2, 3, 2]
index = my_list.index(2)
print(index)
```

### Insert

```python
# Insert an element
my_list = [1, 2, 3]
my_list.insert(0, 0)
print(my_list)
```

```python
# Insert an element
my_list = [1, 2, 3]
my_list.insert(4, 10)
print(my_list)
```

### Pop

```python
# Remove based on the index
my_list = [1, 2, 3]
index = 0
removed_element = my_list.pop(index)
print(removed_element)
print(my_list)
```

### Remove

```python
# Remove an element
my_list = [1, 2, 2, 3]
my_list.remove(2)
my_list
```

### Reverse

```python
# Reverse the order
my_list = [1, 2, 3]
my_list.reverse()
my_list
```

### Sort

```python
# Sort
my_list = [3, 1, 1, 4, 2, 8, 5, 6]
my_list.sort()
my_list
```

```python
# Sort in reverse order
my_list = [3, 1, 1, 4, 2, 8, 5, 6]
my_list.sort(reverse=True)
my_list
```
