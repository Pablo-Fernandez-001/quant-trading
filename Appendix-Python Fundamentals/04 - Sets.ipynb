<h1 style="text-align: justify;">Sets</h1>
<p style="text-align: justify;">

A set is an unordered collection of unique elements. Sets are defined using curly braces {} or the set() function. They support operations such as union, intersection, and difference, which are useful for working with unique data.
    
</p>
</div>

```python
# Create a set
my_set = {1, 2, 3, 4, 5, 5} # Only one 5 will be kept
my_set
```

```python
# Data type
type(my_set)
```

```python
# Create a set with "set"
my_set = set((1, 2, 3, 4, 5, 5))
my_set
```

```python
# Available methods
dir(set)
```

### Add

```python
# Add
example_set = {1, 2, 3}
example_set.add(4)
example_set
```

### Clear

```python
# Clear
my_tuple = (1, 2, 3)
my_set = set(my_tuple)
print(my_set)
my_set.clear()
print(my_set)
```

### Copy

```python
# Create a copy of a set
my_tuple = (1, 2, 3)
my_set = set(my_tuple)
set_copy = my_set.copy()
set_copy
```

### Difference

```python
# Difference
tuple1 = (1, 2, 3, 4)
tuple2 = (3, 4, 5, 6)
set1 = set(tuple1)
set2 = set(tuple2)
difference = set1.difference(set2)
difference
```

### Difference Update

```python
# Update
tuple1 = (1, 2, 3, 4)
tuple2 = (3, 4, 5, 6)
set1 = set(tuple1)
set2 = set(tuple2)
set1.difference_update(set2)
set1 # Updates the first set by removing the elements shared by both sets
```

### Discard

```python
# Remove
tuple1 = (1, 2, 3, 4)
set1 = set(tuple1)
set1.discard(2)
set1
```

### Intersection

```python
# Intersection
tuple1 = (1, 2, 3, 4)
tuple2 = (3, 4, 5, 6)
set1 = set(tuple1)
set2 = set(tuple2)
intersection = set1.intersection(set2)
intersection
```

### Intersection Update

```python
tuple1 = (1, 2, 3, 4)
tuple2 = (3, 4, 5, 6)
set1 = set(tuple1)
set2 = set(tuple2)
set1.intersection_update(set2)
set1
```

### Issubset

```python
# Subset
tuple1 = (1, 2)
tuple2 = (1, 2, 3, 4)
set1 = set(tuple1)
set2 = set(tuple2)
is_subset = set1.issubset(set2)
is_subset
```
